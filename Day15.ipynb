CLASS AND OBJECTS
Class: a blueprint (template) that describes what something has and what it can do. It is not a real thing yet.

Object: a real thing made from that blueprint. You can make many objects from one class.

Example in real life:

Class: the design of a car (it says every car has a brand, color, and can start)
Object: your actual red Toyota, or a black BMW

```python
class Student:                      # class (blueprint)
    def __init__(self, name, marks):
        self.name = name            # data (attribute)
        self.marks = marks

    def show(self):                 # action (method)
        print(self.name, self.marks)

s1 = Student("Moon", 90)            # object
s2 = Student("Anu", 85)             # another object

s1.show()   # Moon 90
s2.show()   # Anu 85
```
