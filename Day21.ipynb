```python
# Safe division
# Take two numbers from the user and print a / b. If b is 0, print "Cannot divide by zero" instead of crashing.
a = int(input("ENter number a : "))
b = int(input("Enter number b : "))
try:
    print(a/b)
except ZeroDivisionError:
    print("Cannot divide by zero")
finally:
    print("Finished exception")
```

```python
# Valid number
# Ask the user for an integer. If they type text like "abc", print "Please enter a number".

try:
    num = int(input("Enter a number : "))
    print(num)
except ValueError:
    print("Please enter number only")
```

```python
# List index
# Given nums = [10, 20, 30], ask for an index and print the value. Handle an index that is out of range.

nums = [10, 20, 30]
try:
    index = int(input("Enter index : "))
    print(nums[index])
except IndexError:
    print("Out of range")
except ValueError:
    print("Name Error")
```

```python
# Dictionary key
# Given d = {"name": "Moon", "age": 22}, ask for a key and print its value. If the key is missing, print "Key not found".

d = {"name": "Vennila", "age": 21}
try :
    key = input("Enter key ")
    print(d[key])
except KeyError:
    print("Key is not found")
```

```python
# User-defined exception

class VennilaException(Exception): #custom exception
    pass

name = input(" enter your name " )

try:    
    if name=="Vennila":
        raise VennilaException(" problem problem problem ")
    else:
        print(name, " proceed ")
except VennilaException as ex:
    print( ex )
```

```python
# Age validation
# Create InvalidAgeError. Write check_age(age) that raises it if age < 0 or age > 120. Take input and handle the error.
class InvalidAgeError(Exception):
    pass
try:
    age = int(input("Enter age : "))
    if age < 0 or age>120:
        raise InvalidAgeError("The age is Invalid")
except InvalidAgeError as iae:
    print(iae)
else:
    print("Your age is ",age)
```

```python
# Minimum balance
# Create MinimumBalanceError. A bank account must keep at least 500. Raise the error if a withdrawal leaves less than 500.

class MinimumBalanceError(Exception):
    pass
    
def withdraw(minimum_balance, account_balance, amount_withdrawal):
    if(account_balance - amount_withdrawal < minimum_balance):
        raise MinimumBalanceError("Withdraw not Successful")
try:
    withdraw(500, 1500, 1300)
except MinimumBalanceError as ex:
    print(ex)
else:
    print("Withdraw successful")
```
