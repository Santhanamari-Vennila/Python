```python
# Sum of digit
num = int(input())
s = 0
while(num>0):
    last_digit = num%10
    s = s+last_digit
    num = num//10
print(s)
```

```python
# Armstrong number
num = int(input()) # 145, 14
temp = num # 145
s = 0 # 0, 125, 189
while(num>0):
    last_digit = num%10  # 5, 4
    s = s + (last_digit**3)  # 0+(5**3)= 125, 125+(4**3) = 189, 
    num = num//10  # 145//10 = 14, 
if s  ==  temp:
    print("true")
else:
     print("false")
```

```python
# Reverse a number
num = int(  input("enter a number") ) # 1234
s = 0 

while num > 0:  
    r = num % 10  
    s=(s*10)+r         
    num=num // 10 
print(" result is " , s)
```

```python
# Palindrome (121 -> 121)
num = int(  input("Enter a number") )
s = 0 
temp = num
while num > 0:  
    r = num % 10 
    s = s+r
    if(num > 9):
        s = s*10
    num=num // 10
if(temp == s):
    print(s," is Palindrome")
else:
    print(s," not a Palindrome")
```

```python
# Repeatedly add all of its digits until the result has only one digit
def my_func(num):
    s = 0
    while(num>0):
        last_digit = num%10
        s = s+last_digit
        num = num//10
    return s
num = int(input())
while(num>9):
    print("my num ", num)
    num = my_func(num)
print(num)
```
