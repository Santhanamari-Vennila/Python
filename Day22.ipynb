Create a class
- Create a class called Student with no attributes or methods.
- Create two objects from that class.
- Print both objects.

```python
class Student:
    def initialize(self,rollno,name,age):
        self.rollno = rollno
        self.name = name
        self.age = age
    def display(self):
        print(f"{self.rollno} | {self.name} | {self.age}")
obj = Student()
obj.initialize(21,"Vennila",21)
obj.display()
```

Instance methods
- Create a class Calculator with two instance variables, a and b.
- Create methods for addition, subtraction, multiplication and division.
- Call all four methods using an object.

```python
class Calculator:
    def __init__(self):
        self.a = int(input("Enter a value : "))
        self.b = int(input("Enter b value : "))
    def addition(self):
        print(f'The addition : {self.a+self.b}')
    def subtraction(self):
        print(f'The subtraction : {self.a-self.b}')
    def multiplication(self):
        print(f'The multiplication : {self.a*self.b}')
    def division(self):
        print(f'The division : {self.a/self.b}')
obj = Calculator()
obj.addition()
obj.subtraction()
obj.multiplication()
obj.division()
```
