```python
# Read marks and print a grade: 90+ is A, 80+ is B, 70+ is C, 60+ is D, otherwise Fail.
grade = int(input("Enter your grade : "))
if(grade >= 90):
    print("90+ is A")
elif(grade >= 80):
    print("80+ is B")
elif(grade >= 70):
    print("70+ is C")
elif(grade >= 60):
    print("60+ is D")   
else:
    print("Fail")
```

```python
# Read a number from 1 to 7 and print the day of the week.
week = input("Enter number from 1 to 7 : ")
match(week):
    case '1':
        print("Sunday")
    case '2':
        print("Monday")
    case '3' :
        print("Tuesday")
    case '4' :
        print("Wednesday")
    case '5' :
        print("Thursday")
    case '6' :
        print("Friday")
    case '7' :
        print("Saturday")
    case _ :
        print("Invalid")
```

```python
# Build a simple calculator: read two numbers and an operator (+, -, *, /) and use match-case.
a = int(input("Enter number a : "))
b = int(input("Enter number b : "))
Op = input("Enter operation : ")
match(Op):
    case '+':
        print(a+b)
    case '-':
        print(a-b)
    case '*':
        print(a*b)
    case '/':
        print(a/b)
```

WHILE LOOP:
    Repeats a block of code as long as the condition is true
Syntax:
while(condition):
    block of code to executed

```python
count = 1
while(count<= 10):
    print(count)
    count +=1
```
