```python
# without multithreading
import time

def function1():
    count = 1
    while count<=10:
        print(" function1: ", count )
        time.sleep(3) # 3 seconds
        count+=1

def function2():
    count = 1
    while count<=10:
        print(" function2: ", count )
        time.sleep(1)
        count+=1
      
function1()
function2()
```

```python
# With multithreading
import time
import threading

def function1():
    count = 1
    while count<=10:
        print(" function1: ", count )
        time.sleep(3) # 3 seconds
        count+=1

def function2():
    count = 1
    while count<=10:
        time.sleep(1)
        print(" function2: ", count ) 
        count+=1

th1=threading.Thread( target = function1 )
th2=threading.Thread( target = function2 )

th1.start()
th2.start()
```

```python
# Print odd and even with two threads
# Thread 1 prints odd numbers, thread 2 prints even numbers, and the output must be in order: 1 2 3 4 5 6 ... up to 10.
import time
import threading

def odd():
    for i in range(1,10):
        if(i%2 == 1):
            print(i)
            time.sleep(1)
def even():
    for i in range(1,10):
        if(i%2 == 0):
            print(i)
            time.sleep(1)

t1 = threading.Thread(target = odd)
t2 = threading.Thread(target = even)

t1.start()
t2.start()
```
