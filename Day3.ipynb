```python
# write a code to print 1 to 20 but ignore the values 5,10,15 using while and if else statements
num = 1
while(num <= 20):
    if(num==5 or num==10 or num==15):
        num +=1
        continue;
    else:
        print(num)
        num+=1
```

```python
n = int(input("Enter a number : "))

if(n%2 == 0):
    if(n>=2 or n<=5):
        print("Not Weird")
    elif(n>=6 or n<=20):
        print("Weird")
    elif(n>20):
        print("Not Weird")
else:
    print("Weird")
```

```python
# Patter question
n = int(input())
for row in range(1,n+1):
    for col in range(1,row+1):
        print("* ",end="")
    print()
```
