Create a class called Student with the following:
Attributes: name, age, and grade
A method display_info() that prints the student's details in a nice format.

```python
class Student:
    def setAttri(self):
        self.name = input("Enter student name : ")
        self.age = int(input("Enter student age : "))
        self.grade = float(input("Enter student grade : "))
    def display_info(self):
        print(f"Student details: \n Student Name:{self.name}\n Student Age: {self.age}\n Student Grade: {self.grade}")
obj = Student()
obj2 = Student()

obj.setAttri()
obj.display_info()

obj2.setAttri()
obj2.display_info()
```

Create a class Rectangle that takes length and width in the constructor.
Add a method area() that returns the area.
Add a method perimeter() that returns the perimeter.
Create a few rectangles and print their area and perimeter.

```python
class Rectangle:
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.width * self.length
        # print(self.width * self.length)
    def perimeter(self):
        return 2*(self.width + self.length)
        # print(2*(self.width + self.length))
obj = Rectangle(2,4)
obj2 = Rectangle(5,10)
obj3 = Rectangle(3,9)

print(obj.area())
print(obj.perimeter())

# obj.area()
# obj.perimeter()
```

```python
# write a code to implement factorial using recursive function
def fact(n):
    res = 1
    for i in range(n,1,-1):
        res *=i
    return res
print(fact(6))
```
