```python
# 1. no parameter, no return type
def addition():# function definition or function implementation
    a=100
    b=200
    c=a+b
    print( c )
addition()
```

```python
# 2. with parameter, no return type
def addition(x,y): # Formal paraameter
    print(x+y)
a = 100
b = 200
addition(a,b) # Actual paramter or original parameters
```

```python
# 3. with parameter, with return type
def addition(x,y):
    return x+y
a = 100
b = 200
result = addition(a,b)
print(result)
```

```python
# 4. no parameter, with return type
def addition():
    a = int(input("Enter a : "))
    b = int(input("Enter b : "))
    c = a+b
    return c
print(addition())
```

```python
# Find odd or even  : no parameter, with return type :
def odd_even():
    n = int(input("Enter a number to find odd or even : "))
    if(n%2 == 0):
        return "Even"
    else:
        return "Odd"
result = odd_even()
print(result)
```

```python
# Write a function to count vowels in a string.
def countVowels():
    vowels = ["a","e","i","o","u"]
    count = 0
    string = input("Enter a string to find vowels ")
    for i in string:
        if i in vowels:
            count += 1
    return count
print(countVowels())
```

```python
# Write a function to find the largest number in a list without using max()
my_list = [10,20,90,40,50]
def largest(my_list):
    largest = 0
    for num in my_list:
        if(largest<num):
            largest = num
    return largest   
print(largest(my_list))
```

```python
# Write a function to check whether a string is a palindrome.

def palindrome(string):
    pos = -1
    reverse = []
    for i in string:
        reverse.insert(pos,i)
        pos -=1
    result = ''.join(reverse)
    print(result)
palindrome(input())
```

```python
# Write a function to return only even numbers from a list.
def even_from_list():
    li = [1,2,3,4,5,6]
    result = []
    for i in li:
        if(i%2 == 0):
            result.append(i)
    print(result)
even_from_list()
```
