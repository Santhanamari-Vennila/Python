CONSTRUCTOR [__init__]:
Constructor is a special method that runs automatically whenever an object is created for that class

```python
# Constructor
class myclass:
    def __init__(self):
        self.a=100
        self.b=200

    def addition(self):
        self.c=self.a + self.b
        print("Addition result is ", self.c)
m1=myclass() # m1={a=100, b=200, c, _init_(), addition() }
m1.addition()
```

```python
# Constructor without parameter
class myclass:
    def __init__(self,x,y):
        self.a=x
        self.b=y

    def addition(self):
        self.c=self.a + self.b
        print("Addition result is ", self.c)
m1=myclass(100,200)
m1.addition()
```

```python
# constructor overloading doesn't support in python
# Which means we cannot have more than one constructor
class myclass:
    def __init__(self):
        self.a=100
        self.b=200
        
    def __init__(self,x,y):
        self.a=x
        self.b=y

    def addition(self):
        self.c=self.a + self.b
        print("Addition result is ", self.c)

m1=myclass()
m2=myclass(100,200)
m1.addition()
m2.addition
# No constructor overloading in Python: Defining a second __init__ method overwrites the first one instead of overloading it.
# for this it will result in error cause we have multiple constructor and python doesn't support it
```

Variable length arguments

```python
def register_user(username, *roles, **details):
    print("User:", username)
    print("Roles:", roles)
    print("Details:", details)

register_user("jdoe", "admin", "editor", city="Chennai", active=True)
# User: jdoe
# Roles: ('admin', 'editor')
# Details: {'city': 'Chennai', 'active': True}
```

```python
class myclass1:
    def initialization(self):
        self.a = 100
        self.b = 50
    def addition(self):
        self.c = self.a + self.b
        print("Addition : ",self.c)
class myclass2(myclass1):
    def subtraction(self):
        self.c = self.a - self.b
        print("Subtraction : ",self.c)
obj = myclass2()
obj.initialization()
obj.addition()
obj.subtraction()
```

```python
class Bank:
    def dodeposit(self):
        print(" code to deposit ")

    def dowithdraw(self):
        print(" code to withdraw ")

    def checkbalance(self):
        print(" code to check balance ")


class NetBank:
    def doRTGS(self):
        print(" code for RTGS ")
    def doNEFT(self):
        print(" code for NEFT ")

class SmartBank(Bank, NetBank):
    def doUPI(self):
        print(" code for UPI ")



obj=SmartBank()

obj.dodeposit()
obj.dowithdraw()
obj.checkbalance()
obj.doRTGS()
obj.doNEFT()
obj.doUPI()
```
