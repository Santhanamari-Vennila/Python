```python
# Fibonacci series : 0, 1, 1, 2, 3, 5, 8, ...
a = 0
b = 1
add = 0
print(a, end = " ")
while(add< 50):
    add = a + b
    print(add, end = " ")
    b = a
    a = add
```

```python
# Geometric Progressions (Powers of X)
# Problem: Calculate the sum of the series \(1 + x + x^2 + x^3 + ... + x^n\)
x = int(input("x = "))
n = int(input("n = "))
sum_ = 0
result = 1
for i in range(0,n):
    result = x**i
    print(result, end = " ")
    sum_ += result
print("\n",sum_)
```

```python
# 2, 5, 11, 23, 47, __
#   3  6  12   27

n = int(input("Enter number of terms: "))
a = 2      
count = 1

while count <= n:
    print(a, end=" ")
    a = a * 2 + 1  
    count += 1       
```

```python
# 1, 2, 6, 24, 120, __
#   1  4  18  96
#  1. 1*1+1 = 2
#  2. 2*2+2 = 6
#  3. 6*3+6 = 24
#  4. 24*4 +24 = 120
#  5. 120*5 + 120 = 720


n = int(input("Enter number of terms: "))
a = 1      
count = 1

while count <= n:
    print(a, end=" ")
    a = a * count + a 
    count += 1       
```
