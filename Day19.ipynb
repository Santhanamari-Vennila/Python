Private variable : cannot be accessed outside the class. If we try to access it it will result in error

```python
class myclass:

    def initialize(self):
        self.__a=100


m1=myclass()  #m1={__a=100, initialize() }

m1.initialize()

print( m1.__a )
```

```python
# name mangling
# ---------------------
#   it is used to access the private variable outside of the class.

class myclass:

    def initialize(self):
        self.__a=100


m1=myclass()  #m1={__a=100, initialize() }

m1.initialize()

print( m1._myclass__a ) #name mangling
```

Polymorphism - same name, different functionality

```python
class bank1:

    def deposit(self):
        print(" code for deposit")

    def withdraw(self):
        print(" code for withdraw")
        
class bank2(bank1):
    pass

b2 = bank2()
b2.deposit()
b2.withdraw()
```

```python
class bank1:

    def deposit(self): # virtual function : it may or may not exist
        print(" code for deposit")

    def withdraw(self): # virtual function
        print(" code for withdraw")



class bank2(bank1):
    def deposit(self): #override : redefinition of the base class function
        print(" advanced code for deposit")


b2=bank2()

b2.deposit()
b2.withdraw()
```

```python
# Create an abstract class Employee with an abstract method salary(). Make Manager and Clerk classes with different salary calculations.
from abc import ABC, abstractmethod
class Employee:
    @abstractmethod
    def salary(self):
        pass
        
class Manager(Employee):
    def salary(self):
        self.a = 80000
        print(self.a, " - Salary of manager")
        
class Clerk(Employee):
    def salary(self):
        self.b = 50000
        print(self.b, " - Salary of clerk")
ins = Clerk()
ins.salary()

ins2 = Manager()
ins2.salary()
```

Question: Shape area calculator

Create a parent class Shape with a virtual method area() that prints "Area not defined".

Create two child classes:

Rectangle (length, breadth): override area() to print length × breadth
Circle (radius): override area() to print 3.14 × radius × radius

Create a list with one Rectangle, one Circle, and one plain Shape. Loop over the list and call area() on each.

Test data: Rectangle(4, 5), Circle(3), Shape()

```python
class Shape:
    def area(self):
        print("Area not defined")
class Rectangle(Shape):
    def __init__(self,length,breadth):
        self.length = length
        self.breadth = breadth
    def area(self):
        print(f'Area of rectangle : {self.breadth * self.breadth}')
class Circle(Shape):
    def __init__(self, r):
        self.r = r
    def area(self):
        print(f'Area of circle: {round(3.14 * self.r * self.r)}')
instance = [Shape(), Rectangle(4, 5), Circle(3)]
for i in instance:
    i.area()
```
