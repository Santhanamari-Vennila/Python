```python
# Count the total repeated letter. Print the maximum letter reapeated and the count of it.
s="aabbbccccddd"
seen = {}
for i in s:
    if i not in seen:
        seen[i] = 1
    else:
        seen[i] += 1
# print(seen)
maxi = 0
key = ''
for k,v in seen.items():
    if maxi < v:
        maxi = v
        key = k
print(maxi,key)
```

```python
# Both variable must match and add
elements1= {"a":10, "b":20, "c":30}
elements2= {"j":10, "b":20, "c":30}

for k,v in elements1.items():
    for i,j in elements2.items():
        if k == i:
            print(f"Key is {k} : Sum is {v+j}")
```

```python
# find maximum sum of three consecutive numbers
# .expected output=13
arr = [1, 1, 2, 3, 8, 1, 3, 2]
max_sum = float('-inf')
for i in range(len(arr)-2):
    current_sum = arr[i] + arr[i+1] + arr[i+2]
    if current_sum > max_sum:
        max_sum = current_sum
print(max_sum)    
```

```python
# Input :
mylist = ["apple", "air", "alpha", "ball", "balloon", "bark", "cat", "car", "candy", "dog"]

# now change the list into a dictionary and add keys with the first letter of each word. 
# output= {'a':["apple", "air", "alpha"], 'b':["ball",........."dog"]}
dict = {}
for i in mylist:
    letter = i[0] 
    if i.startswith(letter):
        if letter not in dict:
            dict[letter] = [i]
        else:
            dict[letter].append(i)
print(dict) 
```
