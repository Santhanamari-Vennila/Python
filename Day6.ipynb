```python
# Keep accepting numbers until the user enters 0, then display their sum.
sum = 0
while True:
    i = int(input())
    sum += i
    if(i==0):
        break
print(sum)
```

```python
# Find the largest and smallest numbers among n inputs.
n = int(input("Enter n :"))
my_list = []
while n>0:
    i = int(input())
    my_list.append(i)
    n = n-1
print("Largest number ",max(my_list))
print("Smallest number ",min(my_list))
```

```python
n = int(input())
my_list = [int(input()) for i in range(n)]
largest = my_list[0]
smallest = my_list[0]
for num in my_list:
    if(largest < num):
        largest = num
    if(smallest>num):
        smallest = num

print("Largest number ",largest)
print("Smallest number ",smallest)
```

```python
# Count how many digits in a number are even and how many are odd.
num = int(input("Enter a number : "))
even = 0
odd = 0
while num>0:
    last_digit = num%10
    if(last_digit%2==0):
        even += 1
    else:
        odd += 1
    num = num//10
print("The even count = ",even,"\nThe odd count = ",odd)
```

```python
n = input("Enter a number : ")
splitted = list(n)
odd = 0
even = 0
for i in splitted:
    num = int(i)
    if(num%2==0):
        even += 1
    else:
        odd += 1
print("Even count ", even)
print("Odd count ", odd)
```
