```python
def is_weekend(day):
    match day.lower():
        case "saturday" | "sunday":
            return True
        case "monday" | "tuesday" | "wednesday" | "thursday" | "friday":
            return False
        case _:
            return "Invalid day"
age = int(input())
if(age>=18):
    print("Eligible")
else:
    print("Not Eligible")
```
