```python
# 1 to 1000 find armstrong number
num = int(input())
spli = len(list(str(num)))
temp = num
result = 0
while(num>0):
    ld = num%10
    result = result + (ld**spli)
    num = num // 10
if(result == temp):
    print("It is a amstrong number")
```

```python
# Palindrome from 100 to 200
num = int(input())
temp = num
result = 0
while num>0:
    ld = num%10
    result += ld
    if(num>9):
        result *= 10
    num = num//10
if(temp == result):
    print(temp, " is palindrome")
```
