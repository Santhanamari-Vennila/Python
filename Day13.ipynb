```python
# Function can be passed as an argument to an another function
def specialFunction(fptr):
    fptr()
def sayHello():
    print("Good Morning")
specialFunction(sayHello)
```

```python
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b

def calculate(func, x, y):
    return func(x, y)

print(calculate(add, 3, 4))
print(calculate(multiply, 3, 4))
```

```python
# Return a function from a function
def specialFunction(fptr):
    def innerFunction():
        print("Hello")
        fptr()
        print("Thank you")
    return innerFunction()

def sayHello():
    print(" Good Morning")
specialFunction(sayHello)
```

```python
def specialFunction(fptr):
    def innerFunction():
        print("Hello")
        fptr()
        print("Thank you")
    return innerFunction()

@specialFunction #Decorator
def sayHello():
    print(" Good Morning")
```

```python
def calculate(funct):
    def add(a,b):
        return a+b
        
    def sub(a,b):
        return a-b
        
    def wrapper():
        c,a,b = funct()
        if c == "add":
            return add(a,b)
        elif c == "sub":
            return sub(a,b)
        else:
            return "invalid operation"
            
    return wrapper
    
@calculate
def funct():
    c = input("Enter the function add or sub : ")
    a = int(input("Enter a : "))
    b = int(input("Enter b : "))
    return c, a, b
print(funct())
```

```python
# 1. Greeting wrapper
# Write a decorator greet that prints "Welcome!" before the function runs and "Goodbye!" after it.
def greeting(goodbye):
    print("Welcome!")
    goodbye()
@greeting
def goodbye():
    print("Goodbye!")
```

```python
# 2. Uppercase decorator
# Write a function get_name() that returns "moon". Use a decorator to make the result uppercase. Expected: MOON.
def upper_func(get_name):
    var = get_name().upper()
    print(var)

@upper_func
def get_name():
    return "moon"
```

```python
# 3. Run count message
# Write a decorator that prints "Function started" and "Function ended" around any function.
def s(e):
    print("Function started")
    e()
@s
def e():
    print("Function ended")
```

```python
# 4. Logger
# Write a decorator log that prints the function name and its arguments, then returns the result.
# Example: add(3, 4) prints Calling add with (3, 4) and returns 7.
def calc(add):
    add(3,4)
@calc
def add(a,b):
    print(a+b)
```

```python
# 5. Double the result
# Write a decorator that doubles the return value of any function. add(2, 3) should give 10.
def calculate(add):
    def double(x,y):
        res = add(x,y)
        return res*2
    return double
@calculate
def add(x,y):
    return x+y
    
x = int(input())
y = int(input())
print(add(x,y))
```
