```python
#required positional argument
def display(a, b):
    print( a, b )


#display()  # error
#display(100)  #error
display(100,200)
# display(100,200,300) #error
```

```python
#default arguments
def display(a=10, b=20):
    print( a, b )


# display()
# display(100)
display(100,200)
```

```python
#named arguments
def display(rollno, name, mark):
    print( ' roll no is ', rollno, '  name is ', name, ' mark is ', mark )
# display('arjun', 100, 1) # roll no is  arjun   name is  100  mark is  1
display(name='arjun', mark=100, rollno=1)
```

```python
# Tuple as result
def display(*a):
    print(a)
display(10,100,299)
```

```python
#variable length of arguments
def display( *a ):
    #print( a )
    for elmnt in a:
        print( elmnt )
#display(10)
# display(100, 200)
display(1000,2000,3000)
```

```python
# Dictionary
def display(**a):
    print(a)
display(rollno= 2, name = 'vennila')
```

```python
def display( **a ):
    #print( a )
    for k in a.keys():
        print(k)
    

# display(rollno=1, name='arjun')
display(rollno=2, name='yogesh', mark=100)
```

Lambda function is an anonymous function, can contain any number of arguments.
one-line function with no name (an anonymous function).
Can contain only one expression

syntax:
    lambda arguments: expression

```python
add = lambda a, b : a+b
print(add(10,20))
```

```python
is_even = lambda n : n%2==0
print(is_even(20))
```

```python
name = lambda a : "My name is " + a
print(name("Vennila"))
```

```python
# Largest of 2
big = lambda a,b : a if a>b else b
print(big(10,20))
```

A function calls itself is called Recursive

```python
def display():
    print('Hello')
    display() # Hello -> Hello
display()
```

```python
def factorial(n):
    if n == 1:
        return 1
    else:
        result = n * factorial(n-1)
    return result
print(factorial(5))
```

```python
# implement odd or even logic using lambda function
odd_even = lambda n: "Even" if n%2==0 else "Odd"
print(odd_even(int(input())))
```

```python
def digit_sum(n):
    if n == 0:       
        return 0
    return (n % 10) + digit_sum(n // 10)

num = int(input("Enter a number: "))
print("Sum of digits:", digit_sum(num))
```
