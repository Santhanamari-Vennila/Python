```python
# Dictionary
elements={"rollno":1, "name":"arjun", "mark":100}
print(elements)
```

```python
# give all keys
for key in elements.keys():
    print(key)
```

```python
# give all values
for value in elements.values():
    print(value)
```

```python
# items() - all (key, value) pairs
for i in elements.items():
    print(i)
```

```python
for k,v in elements.items():
    print("Key is ", k, ", Value is ",v)
```

```python
# d.pop("a") - removes the key and returns its value
print(elements.pop("mark"))
```
