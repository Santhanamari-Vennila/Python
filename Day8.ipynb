```python
# Sum of digit until it comes single digit
num = int(input("Enter number"))
result = 0
while(num>0):
    ld = num%10
    result += ld
    if result>9:
        result*=10
    num = num//10
print(result)
```

```python
# REVERSE A NUMBER
num = int(input())
temp = num
print("Original : ",temp)
result = 0
while(num > 9):
    ld = ld%10
    result = (result*10) + ld
    # if(result>9):
        # result *= 10
    num //= 10
print("Reversed : ",result)
```
