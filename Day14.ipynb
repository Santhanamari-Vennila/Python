List comprehension

```python
elements=[11,12,13,14,15]

result=[ element*10 for element in elements  ]

print( result )
```

```python
result=[ element for element in elements  ]

print( result )
```

```python
result={ element:element*10 for element in elements  }

print( result )
```

```python
result=[ element for element in elements if element%2 == 1 ] 

print( result )
```

The map() function executes a specified function for each item in an iterable. 

The item is sent to the function as a parameter. 

Syntax map(function, iterables)

```python
def add(element):
    return element+element
element=[11,12,13,14,15]
result = list(map(add,element))
print(result)
```

filter() function is used to extract elements from an iterable (like a list, tuple or set) that satisfy a given condition.
Filter only the true data

```python
def process(element):
    if element%2 == 1:
        return True
    else:
        return False
elements=[11,12,13,14,15]
result = list( filter( process, elements) )
print( result )
```

```python
def process(element):
    if element%2 == 0:
        return True
    else:
        return False
elements=[11,12,13,14,15]
result = tuple( filter( process, elements) )
print( result )
```

```python
# Lazy evaluation is used with iterator
assignments=[ [ {'eid':1, 'tasks':['task1','task2', 'task3'] }, 
                {'eid':2, 'tasks':['task1', 'task3'] },
                {'eid':3, 'tasks':['task1','task2', 'task3'] } ], 
              [ {'eid':4, 'tasks':['task1','task2', 'task3'] }, 
                {'eid':5, 'tasks':['task1', 'task3'] },
                {'eid':6, 'tasks':['task1','task2', 'task3']} ], 
              [ {'eid':7, 'tasks':['task1','task2', 'task3'] }, 
                {'eid':8, 'tasks':['task1', 'task3'] },
                {'eid':9, 'tasks':['task1','task2', 'task3'] } ] ]
result = iter( assignments)
print( next(result) )
print(' other codings ....' )
print( next(result) )
print( ' other codings...')
print( next(result) )
```

```python
def process(elements):
    for element in elements:
        if len(element) >= 5:
            yield element


assignments=[ [1,2,3,4,5,6], [11,12,13], [21,22,23,24,25,26,27], [31,32] ]

result=process(assignments)

print( next(result) )

print( next(result) )
```

```python
def long_words(words):
    for w in words:
        if len(w) >= 5:
            yield w

result = list(long_words(["hi", "hello", "python", "sun"]))
print(result)
```
